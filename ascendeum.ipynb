```python
import pandas as pd
import numpy as np


df = pd.read_csv('analysis_id.csv')
print("data loaded successfully")

df.isnull().sum()

df.info()

total_bids = 5000

df['prebid_win_count'] = df['prebid_win_count'].fillna(0)

df.isnull().sum()
# total_wins = df['prebid_win_count'].sum()
total_wins = df['win_count'].sum()

# print(f"Total wins: {total_wins}")  
print(f"Total wins: {total_wins}")  

winner = df[df['win_count'] > 0]

# winner_percentage = (winner['prebid_win_count'] / total_wins) * 100
winner['win_percentage'] = (winner["win_count"] / total_bids) * 100

# highest_percentage = winner.groupby("bidder").agg(
#     highest_wins_bidder = ("win_percentage", "mean")
# ).reset_index().sort_values(by="highest_wins_bidder", ascending=False)
# print(highest_percentage.head(3))  

highest_percentage = winner.groupby("bidder").agg(
    highest_wins_bidder = ("win_percentage", "mean")

).reset_index().sort_values(by="highest_wins_bidder", ascending=False)
print(highest_percentage.head(10))
```
